```python
# Cell 1 – Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/conveyor_belt_data.csv")

print(df.head())
```

```python
# Cell 3 – Dataset information
print(df.shape)
print(df.info())
print(df.describe())
```

```python
# Cell 4 – Check jam records
print(
    df["next_shift_jam_status"].value_counts()
)
```

```python
# Cell 5 – Visualize jam status
df["next_shift_jam_status"].value_counts().plot(
    kind="bar",
    title="Conveyor Jam Status Distribution"
)

plt.xlabel("Jam Status")
plt.ylabel("Number of Records")
plt.show()
```
